# Task 13: Direct Preference Optimization (DPO) and Pairwise Reward Alignment


## Objective

Implement the core DPO loss using chosen/rejected responses and a frozen reference model.


## Short Theory

DPO directly optimizes preference pairs by comparing policy and reference log-probability ratios, avoiding a separate reward-model/RL loop.


## Step 1: Imports


```python
import torch
import torch.nn.functional as F
```

## Step 2: Pairwise Log-Probabilities


```python
policy_chosen = torch.tensor([-2.0, -1.5])
policy_rejected = torch.tensor([-3.0, -2.5])
ref_chosen = torch.tensor([-2.2, -1.8])
ref_rejected = torch.tensor([-2.8, -2.3])
```

## Step 3: DPO Loss


```python
beta = 0.1
chosen_ratio = policy_chosen - ref_chosen
rejected_ratio = policy_rejected - ref_rejected
loss = -F.logsigmoid(beta * (chosen_ratio - rejected_ratio)).mean()
print("DPO loss:", loss.item())
```

## Step 4: Preference Signal


```python
margin = (chosen_ratio - rejected_ratio).mean()
print("Preference margin:", margin.item())
```

## Small Experiment

Pairwise Experiment


```python
print("Chosen preferred:", bool(margin > 0))
```

## Conclusion

Implemented the core pairwise DPO objective using policy/reference relative log-probabilities.
