Kafka provides the stream, Spark Structured Streaming processes micro-batches, sentence-transformers creates vectors, and Qdrant stores/upserts them.


## Step 1: Imports


```python
# pip install sentence-transformers qdrant-client
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
```

## Step 2: Vectorize Stream Payloads


```python
events = [
    {"id": "1", "text": "Transformer attention"},
    {"id": "2", "text": "Vector databases enable semantic search"}
]
texts = [e["text"] for e in events]
vectors = model.encode(texts, normalize_embeddings=True)
print("Vector shape:", vectors.shape)
```
