# Task 1: Vectorized Scaled Dot-Product Attention from Scratch


## Objective

Implement scaled dot-product attention using pure NumPy with batch-wise multi-head 4D tensors and causal masking.


## Short Theory

Attention computes softmax(QKᵀ/√dₖ)V. The causal mask prevents a token from attending to future tokens.


## Step 1: Import Libraries


```python
import numpy as np
np.random.seed(60)
```

## Step 2: Create Batch Multi-Head Q, K and V


```python
batch, heads, seq_len, d_k = 5, 8, 4, 8
Q = np.random.randn(batch, heads, seq_len, d_k)
K = np.random.randn(batch, heads, seq_len, d_k)
V = np.random.randn(batch, heads, seq_len, d_k)
print("Q:", Q.shape, "K:", K.shape, "V:", V.shape)
```

## Step 3: Scaled Attention Scores


```python
scores = Q @ K.transpose(0, 1, 3, 2)
scores = scores / np.sqrt(d_k)
print("Scores:", scores.shape)
```

## Step 4: Apply Causal Mask


```python
mask = np.triu(np.ones((seq_len, seq_len), dtype=bool), k=1)
scores = np.where(mask[None, None, :, :], -1e9, scores)
```

## Step 5: Softmax and Attention Output


```python
e = np.exp(scores - scores.max(axis=-1, keepdims=True))
weights = e / e.sum(axis=-1, keepdims=True)
output = weights @ V
print("Weights:", weights.shape)
print("Output:", output.shape)
print("Row sums:", weights.sum(axis=-1))
```

## Small Experiment

## Conclusion

Pure NumPy 4D scaled dot-product attention was implemented with scaling, multi-head batching, and causal masking.
