# Task 4: In-Memory HNSW Vector Indexing from Scratch


## Objective

Construct a small multi-layer proximity graph and perform cosine-similarity search.


## Short Theory

HNSW organizes vectors into layers and searches from upper layers toward the bottom layer, using graph neighbors to approach high-similarity candidates.


## Step 1: Imports


```python
import numpy as np
import networkx as nx
np.random.seed(42)
```

## Step 2: Create Vectors and Layers


```python
X = np.random.randn(20, 8)
X /= np.linalg.norm(X, axis=1, keepdims=True)
layers = {i: np.random.randint(0, 3) for i in range(len(X))}
G = [nx.Graph() for _ in range(3)]
for i in range(len(X)):
    for l in range(layers[i] + 1):
        G[l].add_node(i)
for l in range(3):
    nodes = list(G[l].nodes)
    for i in nodes:
        sims = X[nodes] @ X[i]
        for j in np.argsort(sims)[-3:]:
            if nodes[j] != i:
                G[l].add_edge(i, nodes[j])
print("Layers:", [len(g.nodes) for g in G])
```

## Step 3: Cosine Search


```python
q = np.random.randn(8); q /= np.linalg.norm(q)
scores = X @ q
top = np.argsort(scores)[-5:][::-1]
print("Top IDs:", top)
print("Similarities:", scores[top])
```

## Small Experiment

Layer Experiment


```python
for j, h in enumerate(G):
    print(f"Layer {j}: nodes={h.number_of_nodes()}, edges={h.number_of_edges()}")
```

## Conclusion

Demonstrated an in-memory layered proximity graph and cosine-similarity retrieval without a vector database driver.
