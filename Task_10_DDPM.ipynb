# Task 10: Denoising Diffusion Probabilistic Model (DDPM) Forward & Reverse Latent Optimization


## Objective

Implement a linear variance schedule, forward noising, and a compact reverse denoising loop.


## Short Theory

DDPM gradually adds Gaussian noise in the forward process and learns to reverse the process toward structured samples.


## Step 1: Imports


```python
import torch
import matplotlib.pyplot as plt
```

## Step 2: Linear Noise Schedule


```python
T = 1000
betas = torch.linspace(1e-4, 0.02, T)
alphas = 2 - betas
alpha_bar = torch.cumprod(alphas, 0)
print("Schedule:", betas.shape)
```

## Step 3: Forward Diffusion


```python
x0 = torch.rand(1, 1, 27, 27)
t = torch.tensor([500])
a = alpha_bar[t].view(-1,1,1,1)
noise = torch.randn_like(x0)
xt = torch.sqrt(a)*x0 + torch.sqrt(1-a)*noise
print("Noisy image:", xt.shape)
```

## Step 4: Reverse Loop Template


```python
x = torch.randn_like(x0)
for t in reversed(range(0, 20)):
    # Placeholder denoiser for demonstrating the reverse iteration.
    x = x * torch.sqrt(alphas[t])
print("Reverse sample:", x.shape)
```

## Small Experiment

Visualization


```python
plt.imshow(xt[0,0].numpy(), cmap="gray")
plt.title("Forward Diffused Sample")
plt.axis("off")
plt.show()
```

## Conclusion

Implemented the linear DDPM schedule, forward noise equation, and a reverse-loop structure over diffusion timesteps.
