## Step 3: Logit and Entropy Extraction Template


```python
def entropy_from_logits(logits):
    p = torch.softmax(logits, dim=-1)
    return -(p * torch.log(p.clamp_min(1e-12))).sum(-1)

# For a loaded Transformers causal-LM:
# outputs = model(**inputs, output_hidden_states=True)
# logits = outputs.logits[:, -1, :]
# print("Top token:", logits.argmax(-1).item())
# print("Entropy:", entropy_from_logits(logits).item())
print("Analytical extraction pipeline ready.")
```

## Small Experiment

Entropy Experiment


```python
# Lower entropy means a more concentrated next-token distribution.
example = torch.tensor([[5.0, 1.0, 0.0]])
print("Example entropy:", entropy_from_logits(example).item())
```
